# MMR (Maximum Marginal Retrival)

It is used to return more diversified results to the user query.

Working: 

1. The first document will be the closest document.
2. the second document will be least similar to the first one . but still it will be similar to the user query. 

this is how MMR adds a touch of diversification.

```python
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings
```

```python
# Sample documents
docs = [
    Document(page_content="LangChain makes it easy to work with LLMs."),
    Document(page_content="LangChain is used to build LLM based applications."),
    Document(page_content="Chroma is used to store and search document embeddings."),
    Document(page_content="Embeddings are vector representations of text."),
    Document(page_content="MMR helps you get diverse results when doing similarity search."),
    Document(page_content="LangChain supports Chroma, FAISS, Pinecone, and more."),
]
```

```python
vector_store = FAISS.from_documents(
    documents = docs,
    embedding = OpenAIEmbeddings()
)
```

```python
retriever = vector_store.as_retriever(
    search_type = 'mmr',
    search_kwargs = {"k":3 , "lambda_mult":0.5}
)
# lambda_mult : 1 normal similarity search, if lambda_mult : 0 the result will be very diversified.
```

```python
query = "What is langchain?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```
