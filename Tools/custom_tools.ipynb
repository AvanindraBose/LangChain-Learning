# Method1 : Using Tool decorator

# Step1 : define the function

```python
def multiply(a,b):
    return a * b
```

# Step2 : Mention the type hints

```python
def multiply(a:float,b:float) -> float:
    return a*b
```

# Step3: Mention the purpose of the fucntion in docstrings

```python
def multiply(a:float,b:float) -> float:
    '''
    The purpose of the function is to multiply the 2 numbers given in the input.
    '''
    return a * b
```

# Step4 : Tool Decorator

```python
@tool
def multiply(a:float,b:float) -> float:
    '''
    The purpose of the function is to multiply the 2 numbers given in the input.
    '''
    return a * b
```

```python
multiply.invoke({'a':2 , 'b':3})
```

```python
multiply.name
```

```python
multiply.description
```

```python
multiply.args
```

This is what LLM will see when it will communicate with the tool :

```python
print(multiply.args_schema.model_json_schema())
```

# @tool should be used when:

You have a simple Python function.

The tool has a small number of straightforward arguments.

You want LangChain to automatically generate the tool schema.

You don't need custom tool lifecycle/error-handling behavior.

You're prototyping or building a normal application tool.

# Method 2: With Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel , Field
```

```python
class inputvariable(BaseModel):
    a : float = Field(required = True ,description="First number to muliply")
    b : float = Field(required = True ,description="Second number to multiply")
```

```python
def multiply_func(a: float, b: float) -> float:
    return a * b
```

```python
multiply_tool = StructuredTool(
    func=multiply_func,
    name="multiply",
    description="Multiply 2 numbers",
    args_schema=inputvariable
)
```

```python
multiply_tool.invoke({'a':5,'b':2})
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Structured Tool should be used when: 

Your tool has multiple structured parameters.

You want strict input validation.

You need a custom Pydantic schema.

The input has constraints, optional fields, nested structures, etc.

You want the tool's schema to be very explicit for the LLM.
