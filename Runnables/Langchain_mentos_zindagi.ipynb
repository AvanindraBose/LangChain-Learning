# New Runnables Logic

```python
class Runnable(ABC):

    @abstractmethod
    def invoke():
        pass
```

```python
class dummyLLM(Runnable):

    def __init__(self):
        print("Initiated LLM")

    def invoke(self,prompt):
        response = [
                    "Virat Kohli is the Best Cricketer.",
                    "Artificial Intelligence is the future.",
                    "India is a very lovable country."
                ]
        
        return {"response":random.choice(response)}


    def predict(self,prompt):
        response = [
            "Virat Kohli is the Best Cricketer.",
            "Artificial Intelligence is the future.",
            "India is a very lovable country."
        ]

        return {"response":random.choice(response)}
```

```python
llm = dummyLLM()
```

```python
llm.invoke("Tell Me about AI")
```

```python
class dummyPromptTemplate(Runnable):

    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self,input_dict):
        return self.template.format(**input_dict)


    def format(self,input_dict:dict):
        return self.template.format(**input_dict)
```

```python
template = dummyPromptTemplate(
    template="Wrtie about {topic}",
    input_variables=["topic"]
)
```

```python
prompt = template.invoke({'topic':'AI'})
```

```python
llm.invoke(prompt)
```

# Chaining Logic in Runnable

```python
class dummyLLMPromptChain(Runnable):

    def __init__(self,runnables:list):
        self.runnables = runnables

    def invoke(self,input_data):

        for run in self.runnables:
            # output of 1 step has to be input of another step
            input_data = run.invoke(input_data)

        final_response = input_data

        return final_response

    def run(self,input_data):
        self.prompt = self.template.format(input_data)
        response = self.llm.predict(self.prompt)
        
        return response['response']
```

```python
chain = dummyLLMPromptChain([template,llm])
```

```python
chain.invoke({'topic':'AI'})
```
