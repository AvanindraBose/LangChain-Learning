```python
import random
```

# Old way of wroking with LLM's without Chains

```python
class dummyLLM:

    def __init__(self):
        print("Initiated LLM")

    def predict(self,prompt):
        response = [
            "Virat Kohli is the Best Cricketer.",
            "Artificial Intelligence is the future.",
            "India is a very lovable country."
        ]

        return {"response":random.choice(response)}
```

```python
class dummyPromptTemplate:

    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables
    
    def format(self,input_dict:dict):
        return self.template.format(**input_dict)
```

```python
llm = dummyLLM()
```

```python
template = dummyPromptTemplate(
    template = "Write about {topic}",
    input_variables = ["AI"]
)
```

```python
prompt = template.format({"topic":"AI"})
```

```python
llm.predict(prompt)
```

# Still Old, but introduction of chains for each and every component

```python
class dummyLLMPromptChain:

    def __init__(self,llm,template):
        self.llm = llm
        self.template = template
    
    def run(self,input_data):
        self.prompt = self.template.format(input_data)
        response = self.llm.predict(self.prompt)
        
        return response['response']
    
```

```python
chain = dummyLLMPromptChain(llm,template)
```

```python
chain.run({"topic":"AI"})
```
